```python
import random
code=str(random.randint(100,999))
print('Crack the code!')
while True:
  guess=input('Enter a 3-digit code:')
  if guess==code:
      print('You cracked the code')
      break
  if guess[0]==code[0]:
      print('First digit is correct')
  if guess[1]==code[1]:
      print('Second digit is correct')
  if guess[2]==code[2]:
      print('Third digit is correct')
  print('Try again!')
```
