```python
import random
n=random.randint(1,100)
guesses=0
a=-1
while a!=n:
  guesses+=1
  a=int(input('Enter a number:'))
  if a>n:
    print('Lower')
  elif a<n:
    print('Higher')
  else:
    print('You guessed it right!')
print(f'The number you guessed in {guesses} guesses is', n)
```
